```python
import pandas as pd
import numpy as np

train_df=pd.read_csv("D:/imdb-sentiment-analysis/data/train.csv")
test_df=pd.read_csv("D:/imdb-sentiment-analysis/data/test.csv")

print(train_df.head())
print(train_df.shape)
print(test_df.head())
print(test_df.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow_vectorizer = CountVectorizer(max_features=5000)

X_train_bow=bow_vectorizer.fit_transform(train_df['review'])
print(X_train_bow.shape)


print(bow_vectorizer.get_feature_names_out()[:20])

X_test_bow=bow_vectorizer.transform(test_df['review'])
print(X_test_bow.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tf_idf_vectorizer=TfidfVectorizer(max_features=5000)

X_train_tf_idf=tf_idf_vectorizer.fit_transform(train_df['review'])
print(X_train_tf_idf.shape)

print(tf_idf_vectorizer.get_feature_names_out()[:20])

X_test_tf_idf=tf_idf_vectorizer.transform(test_df['review'])
print(X_test_tf_idf.shape)
```
